```python
# Importar librerías
import requests
from bs4 import BeautifulSoup
```

```python
# URL base de la página a scrapear
base_url1 = "https://books.toscrape.com/"
base_url =requests.get(base_url1)
html_obtenido = base_url.text
```

```python
# parsear ese HTML
soup = BeautifulSoup(html_obtenido, "html.parser")
```

```python
# Crear una lista para guardar los datos
books_data = []
```

```python
# prompt: extraer de cada producto title, price y instock.

# Encontrar todos los elementos que contienen información de los libros
products = soup.find_all("article", class_="product_pod")
title = []
price = []
instock = []
rating = []

# Iterar sobre cada producto
for product in products:
    # Extraer el título
    titulo = product.h3.a["title"]

    # Extraer el precio
    precio = product.find("p", class_="price_color").text
    precio = float(precio.replace("Â£", ""))  # Eliminar el símbolo de libra y convertir a float
    #(strip=True).replace('$','')
    # Extraer la disponibilidad (instock)
    disponible = product.find("p", class_="instock availability").text.strip()
    if "In stock" in disponible:
        disponible = disponible.replace("In stock", "True")
    else:
        disponible = disponible.replace("Out of stock", "False")

    # Extraer la calificación (icon-star)
    rating_element = product.p["class"][1] # Access the second class, which contains the rating
    # Agregar los datos a la lista
    books_data.append({"title": titulo, "price": precio, "rating": rating_element, "instock": disponible})

# Imprimir los datos extraídos
for book in books_data:
    print(book) # This line needs to be indented to be part of the for loop
```

```python
# prompt: guardar lo base de datos books_data en import csv

import csv

# ... (Your existing code) ...

# Specify the file path where you want to save the CSV file
csv_file_path = 'books_data.csv'

# Write data to a CSV file
try:
    with open(csv_file_path, 'w', newline='', encoding='utf-8') as csvfile:
        fieldnames = ['title', 'price', 'rating', 'instock']  # Define the header row
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

        writer.writeheader()  # Write the header row to the CSV file
        writer.writerows(books_data) # Write the book data to the CSV file

    print(f"Data successfully saved to {csv_file_path}")

except Exception as e:
    print(f"An error occurred while saving the data to CSV: {e}")
```
